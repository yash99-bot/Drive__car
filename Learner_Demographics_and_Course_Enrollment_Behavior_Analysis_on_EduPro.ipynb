```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel(r'EduPro Online Platform (1).xlsx')
data.head()
data.isnull().sum()
data.duplicated().sum()
data.columns
```

```python
data['UserID'].value_counts()
data["Age"].between(0, 15).sum()
data["Age"].between(15, 30).sum()
data["Age"].between(50, 75).sum()
pd.cut(data["Age"], bins=[0, 15, 50, 75], labels=["Low", "Medium", "High"]).value_counts()
```

```python
data['Gender'].unique()
data['Gender'].value_counts()
```
