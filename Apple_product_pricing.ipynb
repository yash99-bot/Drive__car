```python
summary = data.groupby('Platform').agg(Total_Count=('Platform','count'),Total_Sum=('Current_Price_INR','sum')).reset_index()
print(summary)
# intrective table
# Platforms product count and totle current_price
```

```python
summary.set_index('Platform')[['Total_Count', 'Total_Sum']].plot(
    kind='pie',
    subplots=True,        # Required when plotting multiple columns as pie charts
    autopct='%1.1f%%',
    figsize=(12, 6),      # Widened figsize to fit both charts side-by-side
    legend=False          # Hides repetitive legends since slices are labeled
)
# plotting columns palteform
```

```python
sum_difference = summary['Total_Sum'].diff().iloc[-1]

print(f"The total price difference between the platforms is: {abs(sum_difference)}")
```

```python
filtered_df = data[data['Discount_Pct'].between(0,10)]
total_df = filtered_df['Model_Name'].count()
print(total_df)
```

```python
data["Discount_Pct"].between(0, 25).sum()
data["Discount_Pct"].between(25, 50).sum()
data["Discount_Pct"].between(50, 75).sum()
data["Discount_Pct"].between(75, 100).sum()
```

```python
pd.cut(data["Discount_Pct"], bins=[0, 25, 50, 75, 100], labels=["Low", "Medium", "High", "Max"]).value_counts()
```

```python
avg_discount = data["Discount_Pct"].mean()
print(f"Overall Average Discount: {avg_discount:.2f}%")
```

```python
data.groupby(pd.cut(data["Discount_Pct"],bins=[0, 25, 50, 75, 100],labels=["Low", "Medium", "High", "Max"],)).size().plot(kind='bar', color="skyblue",
                                                                                                                          edgecolor="black", rot=0)
```
